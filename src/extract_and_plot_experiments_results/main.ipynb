```python
ORIGIN_PATH = "./.."
```

```python
PATH_GRAPHEIN = "Graphein_notebooks"
PATH_PDB = "PDBGraphStore_notebooks"
```

```python
T = "time"
MEM = "memory"
```

```python
b = "baseline"
gnn = "gnn"
tdc = "tdc"
```

```python
datasets = [b, gnn, tdc]
```

```python
aproaches = [PATH_GRAPHEIN, PATH_PDB]
```

## Workflow

```python
import numpy as np
import pandas as pd
```

```python
mem_df = pd.DataFrame(
    index=["baseline", "tdc", "gnn"],
    columns=["graphein", "pdb"],
    dtype=float,
)

# mapeia o nome da pasta para o nome da coluna
col_map = {
    "Graphein_notebooks": "graphein",
    "PDBGraphStore_notebooks": "pdb",
}

for ap in aproaches:
    for ds in datasets:
        print(f"{ap}/{ds}")

        with open(f"{ORIGIN_PATH}/{ap}/{ds}_{T}.txt", "r") as f:
            s = f.read()
            l = s.split("\n")
            l = [float(i) for i in l]

            avg = np.average(l)
            print(f'time avg (s): {avg}')
            print(f'timestdev: {np.std(l)}')

        mem_l = []

        with open(f"{ORIGIN_PATH}/{ap}/{ds}_{MEM}.txt", "r") as f:
            mem_l = [i.split("/")[0].strip() for i in f.read().split("\n") if i.strip()]

        for i in range(len(mem_l)):
            if "M" in mem_l[i]:
                mem_l[i] = float(mem_l[i].split("M")[0])
            elif "G" in mem_l[i]:
                mem_l[i] = float(mem_l[i].split("G")[0]) * 1000

        mem_df.loc[ds, col_map[ap]] = np.max(mem_l)
        print("\n")

mem_df.head()
```

```python
mem_df["compression_ratio"] = 1 - mem_df['pdb']/mem_df['graphein']
mem_df.head()
```
