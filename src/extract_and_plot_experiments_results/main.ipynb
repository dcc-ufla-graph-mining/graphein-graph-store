```python
ORIGIN_PATH = "./.."
```

```python
PATH_GRAPHEIN = "Graphein_notebooks"
PATH_PDB = "PDBGraphStore_notebooks"
```

```python
T = "time"
MEM = "memory"
```

```python
b = "baseline"
gnn = "gnn"
tdc = "tdc"
```

```python
datasets = [b, gnn, tdc]
```

```python
aproaches = [PATH_GRAPHEIN, PATH_PDB]
```

```python
import numpy as np
```

```python
for ap in aproaches:
    for ds in datasets:
        print(f"{ap}/{ds}")
        with open(f"{ORIGIN_PATH}/{ap}/{ds}_{T}.txt", "r") as f:
            s = f.read()
            l = s.split("\n")
            l = [float(i) for i in l]

            avg = np.average(l)
            print(f'time: {avg}')

        mem_l = []

        with open(f"{ORIGIN_PATH}/{ap}/{ds}_{MEM}.txt", "r") as f:
            mem_l = f.read().split("\n")

            mem_l = [i.split("/")[0] for i in mem_l]

            for i in range(len(mem_l)):
                if "M" in mem_l[i]:
                    mem_l[i] = float(mem_l[i].split("M")[0])
                elif "G" in mem_l[i]:
                    mem_l[i] = float(mem_l[i].split("G")[0]) * 1000

            
            max = np.max(mem_l)
            print(f'memory: {max}')
        print("\n")
```
